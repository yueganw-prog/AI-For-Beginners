````python
import numpy as np
import matplotlib.pyplot as plt

# 班级成绩: 每名学生三门课 [数学, 英语, 科学] 的成绩
scores = {
    'Alice': [88, 92, 79],
    'Bob': [70, 85, 90],
    'Carol': [95, 89, 94],
    'David': [60, 66, 72],
}
SUBJECTS = ['Math', 'English', 'Science']

print('数据已就绪, 学生人数:', len(scores))
````

## 任务 1: 每个学生的平均分

写一个函数 `student_average(scores, name)`, 返回某学生的平均分。
提示: `np.mean(scores[name])` 一行就能搞定。

````python
def student_average(scores, name):
    # TODO: 在这里写你的代码
    mean = np.mean(scores[name])
    return mean


# 验证
print('Alice 平均分:', student_average(scores, 'Alice'))
````

## 任务 2: 每门课的班级平均分

写函数 `subject_average(scores, index)` 返回第 `index` 门课的班级平均分。
提示: 先收集所有学生第 `index` 门课的成绩组成一个列表, 再求 `np.mean`。

````python
def subject_average(scores, index):
    # TODO: 在这里写你的代码
    subject_scores = [score[index] for score in scores.values()]
    mean = np.mean(subject_scores)
    return mean


# 验证
for i, subj in enumerate(SUBJECTS):
    print(f'班级{subj}平均分:', subject_average(scores, i))
````

## 任务 3: 找全班平均分最高的学生

写函数 `best_student(scores)` 返回 `(姓名, 平均分)`。
提示: 复用任务 1 的函数, 遍历所有人找最大值。

````python
def best_student(scores):
    # TODO: 在这里写你的代码
    best_name = max(scores, key=lambda name: student_average(scores, name))
    best_avg = student_average(scores, best_name)
    return best_name, best_avg

name, avg = best_student(scores)
print(f'全班最高: {name}, 平均分 {avg:.1f}')
````

## 任务 4: 画柱状图

用 Matplotlib 画一张柱状图, 展示每个学生的平均分。
